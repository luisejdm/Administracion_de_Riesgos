```python
import numpy as np
```

```python
def TRS(n_bonds, coupons, initial_prices, final_prices, fixed_rate, t, pay_total_return, nominales):
    initial_pos = np.sum(n_bonds * initial_prices)
    final_pos = np.sum(n_bonds * final_prices)
    cupones = np.sum(n_bonds*coupons*nominales/t)
    total_return = final_pos - initial_pos + cupones
    tasa = fixed_rate * initial_pos / t
    if pay_total_return:
        return tasa - total_return
    else:
        return total_return - tasa

n_bonds = np.array([12_000, 58_000, 85_000])
coupons = np.array([0.035, 0.032, 0.04])
initial_prices = np.array([1002.45, 98.75, 80.30])
final_prices = np.array([1003.4, 100.42, 101.81])
fixed_rate = 0.038
t = 2
pay_total_return = False
nominales = np.array([1000, 100, 100])
TRS(n_bonds, coupons, initial_prices, final_prices, fixed_rate, t, pay_total_return, nominales)
```
